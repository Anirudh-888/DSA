# In the array print the Alternate elements

```python
array=[1,2,3,4]
print("The alternate elements in the array")
for i in range(0,len(array),2):
    print(array[i])
```

# Given an array arr[]. The task is to find the largest element and return it.

```python
array=[63,65,84,77,10,3]
max=array[0]
for i in range(len(array)):
    if max<array[i]:
        max=array[i]
print("The maximum element in the array",max)
```

# Linear Search: Finding the element in the array searching linearly

```python
array=[1,5,8,7,4]
print("Linear Search Code")
indexfound=0
found=False
target=int(input("Enter the element to find it in the array"))
for i in range(len(array)):
    if target==array[i]:
        found=True
        indexfound=i
        break
if found is True:
    print(f"The target element {target} is found at {indexfound}")
elif found is False:
    print(f"The target element {target} is not found in the array")
else:
    print("Error in the code!! or in the functioning of the code!!!")
```
